```python
# Arthimetic Operators

x = 10
y = 5

print(x+y)
print(x-y)
print(x*y)
print(type(x/y))

print(x//y)
print(x**y)
```

```python
# Assignment Operator

a = 10
a += 5
a -= 5
a /= 5
a *= 5
a //= 5

print(a)
```

```python
# Comparision Operator

x = 10 
y = 15

print( x == y)
print( x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
#identity operator

x = [10,20]
y = [10,20]
print( x is y)
print(x is not y )
```

```python
# MEMBERSHIP OPERATOR

x = [10,20]
y = 10
print(y in x)
print(y not in x)
```

```python
#logical operator
x = 15
y = 0
print( x and y)
print([] and y)
print(False or y)
```
